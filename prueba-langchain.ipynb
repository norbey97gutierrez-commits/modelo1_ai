# Prueba con LangChain

```python
import os
import getpass
from langchain_openai import AzureChatOpenAI
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv

# Cargamos las variables del sistema
load_dotenv()
```

```python
# Nos aseguramos que las credenciales estan definidas
if not os.environ.get("AZURE_OPENAI_API_KEY"):
    os.environ["AZURE_OPENAI_API_KEY"] = getpass.getpass("Enter API KEY for Azure: ")

# Iniciamos el LLM de Avzure
llm = AzureChatOpenAI(
    api_key=os.environ["AZURE_OPENAI_API_KEY"],
    azure_endpoint=os.environ["AZURE_OPENAI_ENDPOINT"],
    azure_deployment=os.environ["AZURE_OPENAI_DEPLOYMENT_NAME"],
    openai_api_version=os.environ["AZURE_OPENAI_API_VERSION"]
)

# Definimos el pronmpt (plantilla)
template = """
Eres un asistente virtual experto en python
pregunta: como se definen las {variables} en python
"""

# Definimos el objeto del PromptTemplate
plantilla = PromptTemplate(
    input_variables=["variables"],
    template=template
)

# Encadenamos la plantilla directamente al LLM usando el pipe (|)
cadena = plantilla | llm

# Pedimos al usuario que ingrese el tipo de variable que el llm va analizar
tipo_variable = input("Escribe el tipo de variable: ")

# Ejecutamos la cadena con el metodo invoke y le pasamos la variable
resultado = cadena.invoke({"variables": tipo_variable})

respuesta = resultado.content

nombre_archivo = "variables.txt"

with open (nombre_archivo, "w", encoding="utf-8") as archivo:
    archivo.write(respuesta)

print(f"Respuesta guardada exitosamente en: {nombre_archivo}")
```
